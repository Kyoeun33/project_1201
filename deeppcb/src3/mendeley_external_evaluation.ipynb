# Mendeley PCB-Defect v2 외부 평가

기존 ResNet18 모델이 새로운 PCB의 **불량 이미지를 얼마나 놓치는지** 봅니다. 각 이미지의 정답은 `defect`입니다. 이미지 점수는 타일 불량 확률의 최댓값이며, 기존 validation에서 정한 임계값을 적용합니다.

아래 셀은 외부 평가 스크립트가 저장한 파일을 읽습니다. 기본 체크포인트는 실험 C입니다. 다른 실험 결과를 보려면 `RESULT_DIR`를 해당 결과 폴더로 설정하세요.

## 1. 실행 환경과 결과 경로

기존 Python 환경을 선택하고, `deeppcb` 폴더에서 다음 명령을 직접 실행합니다.

```powershell
python -m src3.06_mendeley_external_evaluate
```

데이터 감사만 하려면 `--audit-only --output-dir src3/artifacts/external/mendeley_v2/audit`를 지정합니다. `README_mendeley_external.md`에 다른 체크포인트, 장치, 주석 형식의 옵션이 있습니다.

````python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image as DisplayImage, display

cwd = Path.cwd()
candidates = [cwd] + [
    parent / relative
    for parent in (cwd, *cwd.parents)
    for relative in ("src3", "deeppcb/src3", "data_example/DeepPCB/deeppcb/src3")
]
SRC2 = next((folder for folder in candidates
             if (folder / "mendeley_external_config.json").is_file()), None)
if SRC2 is None:
    raise FileNotFoundError("커널 작업 폴더를 src3, deeppcb 또는 프로젝트 폴더로 설정하세요.")

CONFIG_PATH = SRC2 / "mendeley_external_config.json"
config = json.loads(CONFIG_PATH.read_text(encoding="utf-8-sig"))
RESULT_DIR = (SRC2 / config["output_dir"]).resolve()
# 다른 폴더를 선택하려면 위 줄을 바꿉니다.
# RESULT_DIR = SRC2 / "artifacts/external/mendeley_v2/A"

def read_saved_json(path):
    return json.loads(path.read_text(encoding="utf-8-sig")) if path.is_file() else None

summary = read_saved_json(RESULT_DIR / "summary.json")
print(f"결과 폴더 · {RESULT_DIR}")
if summary is None:
    print("저장된 summary.json이 없습니다. 위 실행 명령으로 결과를 생성하세요.")
else:
    print(f"실행 상태 · {summary['status']}")
    if summary.get("error"):
        print(summary["error"])
    if summary.get("missing_checkpoint_settings"):
        print("확인이 필요한 학습 설정", summary["missing_checkpoint_settings"])

path_keys = ("image_dir", "annotations", "checkpoint")
configured_paths = {key: (SRC2 / config[key]).resolve() for key in path_keys}
display(pd.DataFrame([
    {"설정": key, "기본 경로": str(path), "존재": path.exists()}
    for key, path in configured_paths.items()
]))
if summary:
    display(summary["checkpoint"])
    if summary.get("scoring"):
        display(summary["scoring"])
````

## 2. 데이터와 주석 감사

[Mendeley PCB-Defect v2](https://data.mendeley.com/datasets/vdj74sngvn/2)는 불량 이미지 230장과 회전 박스 주석 2,506개를 공개 기준으로 제시합니다. 6개 클래스는 `missing_pad`, `mouse_bite`, `open_circuit`, `short`, `spur`, `spurious_copper`입니다.

v2의 YOLO OBB TXT 행에는 클래스 번호와 정규화된 네 모서리 좌표가 있습니다. 함께 배포된 v1 COCO JSON은 이미지 ID로 주석을 연결하며 픽셀 좌표 `[x, y, width, height]`를 사용합니다. 아래 표의 수치와 연결 상태는 선택한 로컬 경로를 감사한 결과입니다.

````python
audit = read_saved_json(RESULT_DIR / "audit.json")
if audit:
    display(pd.DataFrame({
        "v2 공개 기준": audit["published_v2_reference"],
        "로컬 관측": audit["observed"],
        "차이": audit["difference_from_published_v2"],
    }))
    display({
        key: audit[key] for key in (
            "image_dir", "annotation_path", "annotation_format", "annotation_status",
            "readable_images", "linked_images", "valid_boxes_linked",
            "classes_with_linked_boxes", "known_normal_images", "issue_counts",
        )
    })
    audit_images = pd.read_csv(RESULT_DIR / "audit_images.csv")
    audit_issues = pd.read_csv(RESULT_DIR / "audit_issues.csv")
    display(audit_images.drop(columns="sha256").head(12))
    print(f"감사 항목 {len(audit_issues)}건 · 전체 목록은 audit_issues.csv")
    if not audit_issues.empty:
        display(audit_issues.head(30))
else:
    print("audit.json이 없습니다. 외부 평가 또는 감사 전용 명령을 실행하세요.")
````

## 3. 학습 전처리 확인

스크립트가 로드한 모델, 타일 크기와 경계 처리, RGB 변환, 크기 조정, 정규화 값을 확인합니다. 추론은 `src3`의 타일·입력·모델 함수를 사용합니다.

````python
if summary and summary.get("preprocessing"):
    for key in ("preprocessing", "tiling", "device"):
        display(pd.Series(summary[key], name=key).to_frame())
else:
    print("체크포인트 전처리 기록이 없습니다. 평가 명령을 실행하고 summary.json을 확인하세요.")
````

## 4. 이미지 단위 결과와 재현율

재현율의 분모는 읽을 수 있는 불량 이미지 수입니다. TP는 불량으로 판정한 수, FN은 정상으로 오판한 수입니다. 모든 이미지를 불량으로 판정하는 기준선도 같은 분모를 사용합니다.

````python
images = None
if summary and summary["status"] == "completed":
    images = pd.read_csv(RESULT_DIR / "image_predictions.csv")
    metrics = summary["metrics"]
    actual = images["ground_truth"].eq("defect")
    detected = actual & images["prediction"].eq("defect")
    missed = actual & images["prediction"].eq("normal")
    comparison = pd.DataFrame([
        {"판정 방식": "ResNet18", "평가 이미지": int(actual.sum()),
         "TP": int(detected.sum()), "FN": int(missed.sum()),
         "불량 재현율": detected.sum() / actual.sum()},
        {"판정 방식": "모두 불량", "평가 이미지": int(actual.sum()),
         "TP": int(actual.sum()), "FN": 0, "불량 재현율": 1.0},
    ])
    display(comparison.style.format({"불량 재현율": "{:.2%}"}))
    print(f"파일 {metrics['total_image_files']}장 · 열리지 않아 분모에서 제외 {metrics['unreadable_excluded']}장")
    display(pd.Series(metrics["score_distribution"], name="최고 타일 점수").to_frame())
    display(pd.Series(summary["unavailable_metrics"], name="계산 불가 사유").to_frame())
else:
    print("완료된 추론 결과가 없습니다.")
````

````python
if images is not None:
    print(f"이미지별 추론 · {len(images)}장")
    with pd.option_context("display.max_rows", None):
        display(images)
    tile_path = RESULT_DIR / "tile_predictions.csv"
    print(f"타일별 좌표와 점수 · {tile_path}")
    display(pd.read_csv(tile_path).head(10))
````

## 5. 임계값과 점수 분포

같은 0~1 점수 축에서 전체 이미지와 검출·미검출 그룹의 분포를 확인합니다. 점선은 기존 validation에서 고정한 임계값입니다.

````python
def show_saved_figure(key):
    if images is None:
        return
    entry = summary.get("figures", {}).get(key)
    path = entry.get("path") if isinstance(entry, dict) else entry
    if path and (RESULT_DIR / path).is_file():
        display(DisplayImage(filename=str(RESULT_DIR / path)))
    else:
        print(entry.get("reason", key) if isinstance(entry, dict) else f"그림 파일 없음 · {key}")

show_saved_figure("score_distribution")
show_saved_figure("detected_vs_missed")
````

## 6. 검출·미검출 이미지와 타일 예시

왼쪽은 원본 이미지입니다. 오른쪽 색상은 타일별 분류 점수, 청록색 윤곽은 최고 점수 타일, 주황색 폴리곤은 Mendeley 원본 정답 주석입니다. 타일의 경계와 정답 주석의 겹침은 정성적으로 살펴봅니다. 주황색 주석은 데이터셋이 제공한 정답 위치입니다.

````python
if images is not None:
    for key, title in (("detected_example", "검출 예시"), ("missed_example", "미검출 예시")):
        print(title)
        show_saved_figure(key)
````

## 7. 해석 범위

미검출 수와 불량 이미지 재현율은 이 양성 코호트에서 고정된 임계값이 어떻게 작동하는지 보여줍니다. 정상 이미지가 없어 TN, FP, specificity, 정상 오탐률을 산출할 수 없습니다. 모두 불량으로 판정하는 기준선도 재현율 100%에 도달하므로, 이 결과만으로 정상·불량 구별 성능을 입증할 수 없습니다.
